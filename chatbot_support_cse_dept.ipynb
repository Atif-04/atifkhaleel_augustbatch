```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

print("API key loaded successfully!")
```

```python
knowledge_base = """
Q: What are the CSE department office timings?
A: The CSE department office is open from 9:30 AM to 4:30 PM on working days.

Q: What is the minimum attendance required for students?
A: Students must maintain at least 75% attendance.

Q: How can I apply for a bonafide certificate?
A: Students can request a bonafide certificate through the department office.

Q: How long does it take to receive a bonafide certificate?
A: A bonafide certificate is normally issued within 3 working days.

Q: Who should I contact for examination-related problems?
A: Students should contact the department examination coordinator or the CSE department office.

Q: How can I report an error in my internal marks?
A: Students should contact the concerned subject faculty member and explain the error.

Q: Where can I find the CSE timetable?
A: The timetable is displayed on the department notice board and shared through the official student communication channel.

Q: What should I bring to a laboratory session?
A: Students should bring their required lab record and any other materials specified by the laboratory faculty.

Q: How can I report a problem with a laboratory computer?
A: Students should inform the laboratory faculty member or lab in-charge.

Q: Can I request an internship letter from the department?
A: Yes, students can request an internship-related letter through the department office.

Q: What should I do if I lose my college ID card?
A: Students should contact the appropriate college office and follow the procedure for obtaining a replacement ID card.

Q: How can I get a duplicate ID card?
A: Students should submit a duplicate ID card request through the appropriate college office.

Q: What should I do if my attendance is below the required percentage?
A: Students with attendance below the required percentage must follow the college attendance rules and contact the department for guidance.

Q: When should students register for semester examinations?
A: Students should complete examination registration before the deadline announced by the college.

Q: How can I contact the CSE department office?
A: Students can contact the CSE department office at 040-12345678 or cse@example.com.

Q: Can I collect an official document on behalf of another student?
A: This depends on the document and the authorization requirements of the department.

Q: How can I request a department-related letter?
A: Students should contact the CSE department office and provide the required details for the letter.
"""

print("Knowledge base created successfully!")
```

```python
system_instruction = f"""
You are a CSE Department Student Support Bot.

Your job is to answer student questions using ONLY the knowledge base provided below.

RULES:
1. Use only information contained in the knowledge base.
2. Do not use outside knowledge.
3. Do not make up or guess information.
4. If the answer is not available in the knowledge base, you must refuse to answer.
5. Remember relevant information from earlier messages in the conversation.
6. Give clear and concise answers.
7. When information includes numbers, dates, times, percentages, durations,
   fees, or contact details, reproduce the information accurately.
8. If the user's question is outside the knowledge base, say EXACTLY:

"I don't have that information in my knowledge base. Please contact the CSE Department Office at 040-12345678 or cse@example.com."

KNOWLEDGE BASE:
{knowledge_base}
"""

print("System instruction created successfully!")
```

```python
chat = client.chats.create(
    model="gemini-3.6-flash",
    config={
        "system_instruction": system_instruction,
        "temperature": 0.2
    }
)

print("CSE Support Bot is ready!")
```

```python
print("======================================")
print("   CSE DEPARTMENT SUPPORT CHATBOT")
print("======================================")
print("Type 'quit' or 'exit' to stop.")
print()

while True:
    user_input = input("You: ")

    if user_input.lower().strip() in ["quit", "exit"]:
        print("Bot: Goodbye!")
        break

    response = chat.send_message(user_input)

    print("Bot:", response.text)
    print()
```
